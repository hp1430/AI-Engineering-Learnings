```python
# imports

from langchain_core.documents import Document
from langchain_community.document_loaders import PyPDFLoader
from pathlib import Path
```

```python
docs_by_hand = [
    Document(
        page_content="Python is a programming language.",
        metadata={
            "source": "intro.txt",
            "level": "beginner"
        }
    ),
    Document(
        page_content="Embeddings turn text into vectors.",
        metadata={
            "source": "intro1.txt",
            "level": "beginner"
        }
    )
]

print(docs_by_hand[0])
```

```python
PDFS = Path("pdfs")

pdf_path = sorted(PDFS.glob("*.pdf"))[0]

print(pdf_path)
print(type(pdf_path))
```

```python
loader = PyPDFLoader(pdf_path)
```

```python
pages = loader.load()
```

```python
print(type(pages))
print(len(pages))
print(pages[0])
```

```python
for i, page in enumerate(pages):
    print(f"Page {i+1}")
    print(page.page_content)
    print("\n")
```
