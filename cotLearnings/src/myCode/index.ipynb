```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

# Initialize the OpenAI client pointing to Google's Gemini endpoint
client = OpenAI(
    api_key=os.getenv("GEMINI_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)

response = client.chat.completions.create(
    model="gemini-3.5-flash-lite",
    messages=[
        {"role": "user", "content": "The cafeteria had 23 apples. If they used 20 to make lunch and bought 6 more, how many apples do they have? Answer in a single number. without reasoning."}
    ],
)

print(response.choices[0].message.content)
```

```python
def get_answer_from_gemini(prompt: str) -> str:
    response = client.chat.completions.create(
        model="gemini-3.5-flash-lite",
        messages=[
            {"role": "user", "content": prompt}
        ],
    )
    return response.choices[0].message.content
```

```python
ZERO_SHOT_COT_SUFFIX = "Let’s think step by step"

QUESTION = "The cafeteria had 23 apples. If they used 20 to make lunch and bought 6 more, how many apples do they have? Answer in a single number. without reasoning."

ZERO_SHOT_PROMPT = f"""
Question: {QUESTION}
{ZERO_SHOT_COT_SUFFIX}
"""
```

```python
print(get_answer_from_gemini(ZERO_SHOT_PROMPT))
```

```python
FEW_SHOT_COT_EXAMPLES = """
    Q: There were 9 cars parked in the garage. Lilian parked 3 more cars in the garage. How many cars are in the garage now?
A: There were 9 cars initially. Lilian parked 3 more, so there are 9 + 3 = 12 cars in the garage now. #### 12

Q: Leah had 32 chocolates and her sister had 42. If they ate 35 chocolates in total, how many do they have left?
A: Leah and her sister had a total of 32 + 42 = 74 chocolates. They ate 35 chocolates, so they have 74 - 35 = 39 chocolates left. #### 39

Q: The cafeteria had 23 apples. If they used 20 to make lunch and bought 6 more, how many apples do they have?
A: The cafeteria started with 23 apples. They used 20 to make lunch, leaving 23 - 20 = 3 apples. Then they bought 6 more, which means they now have 3 + 6 = 9 apples. #### 9
"""

FEW_SHOT_PROMPT = f"""
{FEW_SHOT_COT_EXAMPLES}

Question: {QUESTION}

provide me answer as described in the examples above. Let's think step by step without reasoning.
"""

print(get_answer_from_gemini(FEW_SHOT_PROMPT))
```
